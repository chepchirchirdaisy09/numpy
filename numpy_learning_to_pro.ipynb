## DOT PRODUCT EXAMPLE(NUT BUTTER SALES)

```python
 import numpy as np
np.random.seed(0)
sales_amount=np.random.randint(20,size=(5,3))
sales_amount
sales_amount.shape
```

```python
#weekly sales dataframe
import pandas as pd
weekly_sales=pd.DataFrame(sales_amount,index=["mon","tue","wed","thur","fri"],
                         columns=["almond butter","peanut butter","cashew butter"])
weekly_sales
```

```python
#create prices array
import numpy as np
prices=np.array([10,8,12])
prices.shape
```

```python
#create butter_prices dataframes
import pandas as pd
butter_prices=pd.DataFrame(prices.reshape(1,3),
                           index=["prices"],
                           columns=["almond butter","peanut butter","cashew butter"])

butter_prices
```

```python
total_sales=prices.dot(sales_amount)
```

```python
#shapes arent aligned let us transpose
sales_amount.T
sales_amount.T
```

```python

sales_amount
```

```python
total_sales=prices.dot(sales_amount.T)
total_sales
```

```python
#create daily-sales
butter_prices.shape
```

```python
weekly_sales
```

```python
daily_sales=butter_prices.dot(weekly_sales.T)
daily_sales
```

```python
#doesnt work not the right shape
weekly_sales["Total_sales ($)"]=daily_sales.T
weekly_sales
```
